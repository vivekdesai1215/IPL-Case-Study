### Task 4

```python
unique_bowlers = np.unique(bowler)
np.count_nonzero(unique_bowlers)
```

```python
result_4=[]

for one_bowler in unique_bowlers:
    mask = (bowler==one_bowler)
    total_over = np.sum(over[mask])
    total_runss = np.sum(total_runs[mask])
    if total_over != 0:
        economy = total_runss / total_over
        result_4.append((str(one_bowler), float(economy)))
```

```python
unique_batters = np.unique(batter)
```

```python
def runs_per_over(given_over, batsman_name):
    result_5=[]
    mask_batsman = (batter == batsman_name)
    mask_over = (over == given_over)
    final_mask = mask_batsman & mask_over
    runs = np.sum(total_runs[final_mask])
    overs = np.sum(over[final_mask])
    result_5.append((str(batsman_name), float(runs/overs)))

    return result_5
```

### Task 9
- Determine match winner based on total runs scored by each team

```python
unique_matches = np.unique(match_id)
```

```python
result = []

for match in unique_matches:
    match_mask = (match_id == match)

    teams = np.unique(batting_team[match_mask])

    if len(teams) == 2:
        team1 = teams[0]
        team2 = teams[1]

        # Runs scored by team1
        mask_team1 = match_mask & (batting_team == team1)
        runs_by_team1 = np.sum(total_runs[mask_team1])

        # Runs scored by team2
        mask_team2 = match_mask & (batting_team == team2)
        runs_by_team2 = np.sum(total_runs[mask_team2])

        if runs_by_team1 > runs_by_team2:
            winner = team1
        else:
            winner = team2

        result.append((int(match), str(winner)))

print(result[:5])
```

### TASK 10
- Check whether toss-winning team scores more runs

```python
import csv
with open("matches.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)  
    data = [row for row in reader]
 
matches = np.array(data, dtype=str)
```

```python
match_id_match = matches[:, 0]
toss_winner = matches[:, 9]
```

```python
result_10 = []

for i in range(len(match_id_match)):
    current_match_id = int(match_id_match[i])
    toss_team = toss_winner[i]

    match_mask = (match_id == current_match_id)

    teams = np.unique(batting_team[match_mask])

    if len(teams) == 2:
        team1 = teams[0]
        team2 = teams[1]

        # runs scored by team1
        mask_team1 = match_mask & (batting_team == team1)
        runs_team1 = np.sum(batsman_runs[mask_team1])

        # runs scored by team2
        mask_team2 = match_mask & (batting_team == team2)
        runs_team2 = np.sum(batsman_runs[mask_team2])

        # identify opponent of toss winner
        if toss_team == team1:
            toss_winner_runs = runs_team1
            opponent_runs = runs_team2
        elif toss_team == team2:
            toss_winner_runs = runs_team2
            opponent_runs = runs_team1
        else:
            continue


        if toss_winner_runs > opponent_runs:
            impact = "Yes"
        else:
            impact = "No"

        result_10.append((
            int(current_match_id),
            str(toss_team),
            int(toss_winner_runs),
            int(opponent_runs),
            impact
        ))

```

```python
for item in result_10[:10]:
    print(item)
```
