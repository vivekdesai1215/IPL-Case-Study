## **Stage 2**
- Data Cleaning & Validation

```python
print(deliveries_df.isnull().sum())
```

```python
print(deliveries_df['batting_team'].unique())
print(deliveries_df['dismissal_kind'].unique())
```

```python
deliveries_df['dismissal_kind'] = deliveries_df['dismissal_kind'].replace({
    'run out ': 'run out',
    'NA': None
})
```

```python
print(deliveries_df.dtypes)
```

```python
deliveries_df['extras_type'] = deliveries_df['extras_type'].fillna('No Extra')

deliveries_df['player_dismissed'] = deliveries_df['player_dismissed'].fillna('Not Out')

deliveries_df['dismissal_kind'] = deliveries_df['dismissal_kind'].fillna('Not Dismissed')

deliveries_df['fielder'] = deliveries_df['fielder'].fillna('No Fielder Involved')
```

```python
print(deliveries_df.isnull().sum())
```

## **Stage 3**
- Data Transformation

```python
deliveries_df["ball_id"] = (
    deliveries_df["match_id"].astype(str) + "_" +
    deliveries_df["inning"].astype(str) + "_" +
    deliveries_df["over"].astype(str) + "_" +
    deliveries_df["ball"].astype(str)
)
```

```python
deliveries_df["is_boundary"] = deliveries_df["batsman_runs"].apply(
    lambda x: 1 if x in [4, 6] else 0
)
```

```python
deliveries_df["is_four"] = (deliveries_df["batsman_runs"] == 4).astype(int)

deliveries_df["is_six"] = (deliveries_df["batsman_runs"] == 6).astype(int)
```

```python
deliveries_df["is_dot_ball"] = (deliveries_df["total_runs"] == 0).astype(int)
```

```python
matches_df["date"] = pd.to_datetime(matches_df["date"])
```

```python
deliveries_df.columns = deliveries_df.columns.str.lower().str.strip().str.replace(" ","_")
matches_df.columns = matches_df.columns.str.lower().str.strip().str.replace(" ","_")
```

## ***Stage 4***
- Core Analysis

#### Total Runs per Match

```python
total_runs_match = deliveries_df.groupby("match_id")["total_runs"].sum()
print(total_runs_match.head())
```

Runs Per Team Per Match

```python
team_match_runs = deliveries_df.groupby(
    ["match_id", "batting_team"]
)["total_runs"].sum().reset_index()

print(team_match_runs.head())
```

Top 10 Batters

```python
top_batters = deliveries_df.groupby("batter")["batsman_runs"]\
.sum().sort_values(ascending=False).head(10)

print(top_batters)
```

Strike Rate of Batters

```python
strike_rate = deliveries_df.groupby("batter").agg({
    "batsman_runs":"sum",
    "ball":"count"
})

strike_rate["strike_rate"] = (
    strike_rate["batsman_runs"] / strike_rate["ball"]
) * 100

print(strike_rate.sort_values("strike_rate", ascending=False).head(10))
```

Top 10 Bowlers by Economy

```python
bowler_stats = deliveries_df.groupby("bowler").agg({
    "total_runs":"sum",
    "ball":"count"
})

bowler_stats["overs"] = bowler_stats["ball"] / 6
bowler_stats["economy"] = (
    bowler_stats["total_runs"] / bowler_stats["overs"]
)

print(bowler_stats.sort_values("economy").head(10))
```

Most Consistent Batters

```python
batter_match_runs = deliveries_df.groupby(
    ["batter","match_id"]
)["batsman_runs"].sum().reset_index()

consistency = batter_match_runs.groupby("batter").agg({
    "batsman_runs":"mean",
    "match_id":"nunique"
})

consistent_batters = consistency[
    consistency["match_id"] >= 20
].sort_values("batsman_runs", ascending=False)

print(consistent_batters.head(10))
```

Highest Individual Score in a Match

```python
highest_score = deliveries_df.groupby(
    ["match_id","batter"]
)["batsman_runs"].sum().reset_index()

print(highest_score.sort_values(
    "batsman_runs",
    ascending=False
).head(1))
```

Boundary Analysis

```python
total_fours = (deliveries_df["batsman_runs"] == 4).sum()
total_sixes = (deliveries_df["batsman_runs"] == 6).sum()

print("Total Fours:", total_fours)
print("Total Sixes:", total_sixes)
```

Boundary Percentage

```python
boundary_runs = deliveries_df[
    deliveries_df["batsman_runs"].isin([4,6])
].groupby("batter")["batsman_runs"].sum()

total_runs = deliveries_df.groupby("batter")["batsman_runs"].sum()

boundary_percentage = (
    boundary_runs / total_runs
) * 100

print(boundary_percentage.sort_values(
    ascending=False
).head(10))
```

Dot Ball Analysis

```python
dot_balls = deliveries_df[
    deliveries_df["total_runs"] == 0
]

top_dot_bowlers = dot_balls.groupby(
    "bowler"
).size().sort_values(
    ascending=False
).head(10)

print(top_dot_bowlers)
```

Runs Per Over Analysis

```python
runs_per_over = deliveries_df.groupby(
    "over"
)["total_runs"].mean()

print(runs_per_over)
```

Powerplay Analysis (1–6)

```python
powerplay = deliveries_df[
    deliveries_df["over"] <= 6
]

powerplay_runs = powerplay.groupby(
    "batting_team"
)["total_runs"].sum().sort_values(
    ascending=False
)

print(powerplay_runs.head(10))
```

Death Over Analysis (16–20)

```python
death_overs = deliveries_df[
    deliveries_df["over"] >= 16
]

death_runs = death_overs.groupby(
    "batting_team"
)["total_runs"].sum().sort_values(
    ascending=False
)

death_batters = death_overs.groupby(
    "batter"
)["batsman_runs"].sum().sort_values(
    ascending=False
)

print(death_runs.head())
print(death_batters.head())
```

Run Distribution Per Inning

```python
inning_runs = deliveries_df.groupby(
    "inning"
)["total_runs"].mean()

print(inning_runs)
```

```python
merged_df = deliveries_df.merge(
    matches_df,
    left_on="match_id",
    right_on="id"
)
```

Season-wise Trend Example

```python
season_runs = merged_df.groupby(
    "season"
)["total_runs"].sum()

print(season_runs)
```

Toss Impact Analysis

```python
team_runs = deliveries_df.groupby(
    ["match_id", "batting_team"]
)["total_runs"].sum().reset_index()

toss_analysis = team_runs.merge(
    matches_df[["id", "toss_winner"]],
    left_on="match_id",
    right_on="id"
)

toss_analysis["is_toss_winner"] = (
    toss_analysis["batting_team"] == toss_analysis["toss_winner"]
)

result = toss_analysis.groupby(
    "is_toss_winner"
)["total_runs"].mean()

print(result)
```

```python
toss_win_advantage = (
    matches_df["toss_winner"] == matches_df["winner"]
).mean() * 100

print("Toss winner also won match:", toss_win_advantage, "%")
```

Player of Match Contribution

```python
player_runs = deliveries_df.groupby(
    ["match_id", "batter"]
)["batsman_runs"].sum().reset_index()

highest_scorers = player_runs.loc[
    player_runs.groupby("match_id")["batsman_runs"].idxmax()
]

pom_analysis = highest_scorers.merge(
    matches_df[["id", "player_of_match"]],
    left_on="match_id",
    right_on="id"
)

pom_analysis["was_top_scorer"] = (
    pom_analysis["batter"] == pom_analysis["player_of_match"]
)

print(
    pom_analysis["was_top_scorer"].value_counts()
)
```

Winning Team Analysis

```python
innings_runs = deliveries_df.groupby(
    ["match_id", "inning", "batting_team"]
)["total_runs"].sum().reset_index()
```

```python
predicted_winners = []

for match in innings_runs["match_id"].unique():
    match_data = innings_runs[
        innings_runs["match_id"] == match
    ]
    
    if len(match_data) == 2:
        team1 = match_data.iloc[0]
        team2 = match_data.iloc[1]
        
        winner = (
            team1["batting_team"]
            if team1["total_runs"] > team2["total_runs"]
            else team2["batting_team"]
        )
        
        predicted_winners.append({
            "match_id": match,
            "predicted_winner": winner
        })
```

```python
predicted_df = pd.DataFrame(predicted_winners)
```

```python
final_winner_analysis = predicted_df.merge(
    matches_df[["id", "winner"]],
    left_on="match_id",
    right_on="id"
)

final_winner_analysis["correct_prediction"] = (
    final_winner_analysis["predicted_winner"] ==
    final_winner_analysis["winner"]
)

accuracy = (
    final_winner_analysis["correct_prediction"]
    .mean() * 100
)

print("Winner prediction accuracy:", accuracy, "%")
```
